```python
import numpy as np
import pandas as pd
import ast
import warnings
```

```python
warnings.filterwarnings("ignore")
```

```python
mv=pd.read_csv("tmdb_5000_movies.csv")
cr=pd.read_csv("tmdb_5000_credits.csv")
```

```python
mv.head()
```

```python
cr.head(1)
```

```python
mv=mv.merge(cr,on="title")
```

```python
mv.head(1)
```

```python
mv=mv[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
mv.head()
```

```python
mv.isnull().sum()
```

```python
mv.dropna(inplace=True)
```

```python
mv.duplicated().sum()
```

```python
mv.iloc[0].genres
```

```python
def convert(a):
    ls=[]
    for i in ast.literal_eval(a):
        ls.append(i["name"])
    return ls
```

```python
mv["genres"]=mv["genres"].apply(convert)
```

```python
mv.head()
```

```python
mv.iloc[0].keywords
```

```python
mv["keywords"]=mv["keywords"].apply(convert)
```

```python
def convert2(a):
    n=0
    ls=[]
    for i in ast.literal_eval(a):
        if n!=3:
            ls.append(i["name"])
            n+=1
        else:
            break
    return ls
```

```python
mv["cast"]=mv["cast"].apply(convert2)
```

```python
mv.head()
```

```python
def convert3(a):
    ls=[]
    for i in ast.literal_eval(a):
        if i['job']=="Director":
            ls.append(i["name"])
    return ls
```

```python
mv["crew"]=mv["crew"].apply(convert3)
```

```python
mv.head()
```

```python
mv["overview"]=mv["overview"].apply(lambda x: x.split())
```

```python
mv["genres"]=mv["genres"].apply(lambda x:[i.replace(" ","") for i in x])
mv["keywords"]=mv["keywords"].apply(lambda x:[i.replace(" ","") for i in x])
mv["crew"]=mv["crew"].apply(lambda x:[i.replace(" ","") for i in x])
mv["cast"]=mv["cast"].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
mv["tags"]=mv["overview"]+mv["genres"]+mv["keywords"]+mv["crew"]+mv["cast"]
```

```python
mv.head()
```

```python
newdf=mv[['movie_id','title','tags']]
```

```python
newdf
```

```python
newdf["tags"]=newdf["tags"].apply(lambda x: (" ").join(x))
```

```python
newdf["tags"]=newdf["tags"].apply(lambda x: x.lower())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words="english")
```

```python
vectors=cv.fit_transform(newdf["tags"]).toarray()
```

```python
vectors[0]
```

```python
ft=cv.get_feature_names_out()
ft[0:5]
#for i in ft:
  #  print(i)
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
ps.stem("loved")
```

```python
def stem(a):
    ls=[]
    for i in a.split():
        ls.append(ps.stem(i))
    return " ".join(ls)
```

```python
newdf['tags']=newdf['tags'].apply(stem)
```

```python
vectors=cv.fit_transform(newdf["tags"]).toarray()
```

```python
ft=cv.get_feature_names_out()
a=0
for i in ft:
    print(i)
    a+=1
    if a==50:
        break
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(vectors)
```

```python
def recommend(movie):
    movie_index=newdf[newdf["title"]==movie].index[0]
    distances=similarity[movie_index]
    movie_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x: x[1])[1:6]

    for i in movie_list:
        print(newdf.iloc[i[0]].title)  
```

```python
recommend("")
```
